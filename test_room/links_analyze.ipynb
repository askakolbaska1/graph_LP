```python
from platform import node

import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np
import torch
import torch.nn.functional as F
```

```python
df = pd.read_csv('../data/edges/clean_edges_without_NaNm.csv')

G_iw = nx.from_pandas_edgelist(df[df['predicate'] == 'interacts_with'], source='id_entity_1', target='id_entity_2')
G_hs = nx.from_pandas_edgelist(df[df['predicate'] == 'has_similarity'], source='id_entity_1', target='id_entity_2')

degrees_iw = nx.degree_centrality(G_iw)
pr_iw = nx.algorithms.pagerank(G_iw)
cluster_iw = nx.clustering(G_iw)
degrees_hs = nx.degree_centrality(G_hs)
pr_hs = nx.algorithms.pagerank(G_hs)
cluster_hs = nx.clustering(G_hs)

all_nodes = set(G_iw.nodes()) | set(G_hs.nodes())

node_topol_data = {}
for node in all_nodes:
    # Создаем сырые признаки
    raw_features = torch.tensor([
        degrees_iw.get(node, 0),
        pr_iw.get(node, 0),
        cluster_iw.get(node, 0),
        degrees_hs.get(node, 0),
        pr_hs.get(node, 0),
        cluster_hs.get(node, 0),
    ], dtype=torch.float)

    # Тут же нормализуем
    normalized_features = F.normalize(raw_features, p=2, dim=0)
    node_topol_data[node] = normalized_features
```

```python
degrees_iw = np.array([d for n, d in G_iw.degree()])
degrees_hs = np.array([d for n, d in G_hs.degree()])

print('=======================IW=======================')
print(f'Всего связей: {len(G_iw.edges())}')
print(f'Всего узлов: {len(G_iw.nodes())}')
print(f'Максимальная степень узла: {degrees_iw.max()}')
print(f'Узлов со степенью больше 500: {len(degrees_iw[degrees_iw > 500])}')
print(f'50 перцентиль: {np.percentile(degrees_iw, 50)}')
print(f'75 перцентиль: {np.percentile(degrees_iw, 75)}')
print(f'95 перцентиль: {np.percentile(degrees_iw, 95)}')
print(f'99 перцентиль: {np.percentile(degrees_iw, 99)}')

print('=======================HS=======================')
print(f'Всего связей: {len(G_hs.edges())}')
print(f'Всего узлов: {len(G_hs.nodes())}')
print(f'Максимальная степень узла: {degrees_hs.max()}')
print(f'Узлов со степенью больше 500: {len(degrees_hs[degrees_hs > 500])}')
print(f'50 перцентиль: {np.percentile(degrees_hs, 50)}')
print(f'75 перцентиль: {np.percentile(degrees_hs, 75)}')
print(f'95 перцентиль: {np.percentile(degrees_hs, 95)}')
print(f'99 перцентиль: {np.percentile(degrees_hs, 99)}')
```
