```python
import networkx as nx
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('../data/edges/clean_triples.csv')

# 1. Создаем граф из вашего DataFrame
# Мы используем только столбцы с ID для экономии памяти при построении индекса
G = nx.from_pandas_edgelist(df, source='id_entity_1', target='id_entity_2')

# 2. Находим узлы, входящие в самую большую компоненту (LCC)
# nx.connected_components возвращает генератор множеств узлов
lcc_nodes = max(nx.connected_components(G), key=len)

# 3. Фильтруем исходный DataFrame
# Оставляем только те строки, где ОБА объекта (субъект и объект) принадлежат LCC
# Это гарантирует, что все ребра в новом DF лежат внутри одной компоненты
df_lcc = df[df['id_entity_1'].isin(lcc_nodes) & df['id_entity_2'].isin(lcc_nodes)].copy()

print(f"Было ребер: {len(df)}")
print(f"Осталось в LCC: {len(df_lcc)}")
```

```python
df_lcc.to_csv('../data/edges/LargestCC.csv', index=False)
```

```python
components = sorted(nx.connected_components(G), key=len, reverse=True)
lcc_nodes = components[0]
LCC = G.subgraph(lcc_nodes).copy()
print(f"Узлов в LCC: {LCC.number_of_nodes()}")
```
