```python
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv('../data/edges/clean_edges_without_NaNm.csv')
```

```python
df_iw = df[df['predicate'] == 'interacts_with']
df_hs = df[df['predicate'] == 'has_similarity']

G_iw = nx.from_pandas_edgelist(df_iw, source='id_entity_1', target='id_entity_2')
G_hs = nx.from_pandas_edgelist(df_hs, source='id_entity_1', target='id_entity_2')
```
