```python
import pandas as pd
import pickle
```

```python
df = pd.read_csv('../data/edges/clean_triples.csv')
```

```python
df_1 = df[['id_entity_1', 'type_entity_1']].rename(columns={'id_entity_1': 'id_entity', 'type_entity_1': 'type_entity'},)
df_2 = df[['id_entity_2', 'type_entity_2']].rename(columns={'id_entity_2': 'id_entity', 'type_entity_2': 'type_entity'},)
clean_df = pd.concat([df_1, df_2], ignore_index=True).drop_duplicates()
```

```python
for i in clean_df['type_entity'].unique():
    temp_df = clean_df[clean_df['type_entity'] == i]
    temp_df.to_csv(f'../data/nodes/clean_nodes_in_graph/clean_{i}.csv', index=False)
```

```python
df_dna = pd.read_csv('../data/nodes/clean_nodes_in_graph/clean_DNA.csv')
df_NA = pd.read_csv('../data/nodes/clean_nodes_in_graph/clean_NucleicAmbigous.csv')
df_NM = pd.read_csv('../data/nodes/clean_nodes_in_graph/clean_NucleicMixed.csv')

full_dna_df = pd.concat([df_NM, df_NA, df_dna], ignore_index=True)
full_dna_df.to_csv('../data/nodes/clean_nodes_in_graph/clean_DNA.csv', index=False)
```

```python
with open('../data/dicts/bert_embeddings/dnabert_DNA.pkl', 'rb') as f:
    dict1 = pickle.load(f)

with open('../data/dicts/bert_embeddings/dnabert_NucleicMixed.pkl', 'rb') as f:
    dict2 = pickle.load(f)

with open('../data/dicts/bert_embeddings/dnabert_NucleicAmbigous.pkl', 'rb') as f:
    dict3 = pickle.load(f)

combined_dict = dict1 | dict2 | dict3
with open('../data/dicts/bert_embeddings/dna_full.pkl', 'wb') as f:
    pickle.dump(combined_dict, f)
```
